```python
import subprocess, pathlib, datetime as dt
```

```python
input_folder = 'input/adencamera_aden-camera-photofinishing-order-1735985_2026-10-25_1455/00189596'

# start = dt.datetime.strptime(input("Date of NEWEST photo (YYYY-MM-DD HH:MM): "), "%Y-%m-%d %H:%M")
start = dt.datetime.strptime('2026-10-27 16:00', "%Y-%m-%d %H:%M")
mins  = 1

# Toronto coordinates
latitude = "43.6532"
latitude_ref = "N"
longitude = "79.3832"
longitude_ref = "W"
```

```python
files = sorted(p for p in pathlib.Path(input_folder).iterdir() if p.suffix.lower() in (".jpg", ".jpeg", ".tif", ".tiff"))
for i, f in enumerate(files):
    d = (start - i * dt.timedelta(minutes=mins)).strftime("%Y:%m:%d %H:%M:%S")
    subprocess.run(["exiftool", "-overwrite_original", f"-AllDates={d}", f"-FileModifyDate={d}", f"-GPSLatitude={latitude}", f"-GPSLatitudeRef={latitude_ref}", f"-GPSLongitude={longitude}", f"-GPSLongitudeRef={longitude_ref}", "-XMP:Location=Toronto", "-XMP:City=Toronto", "-XMP:Country=Canada", "-XMP:CountryCode=CA", str(f)], check=True)
    print(f.name, "→", d, "→ Toronto")
```
